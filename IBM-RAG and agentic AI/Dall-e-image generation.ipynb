```python
import requests
from PIL import Image
from IPython.display import display

prompt = "A white siamese cat"

# Use the correct image endpoint with proper headers
url = f"https://image.pollinations.ai/prompt/{prompt.replace(' ', '%20')}"

headers = {
    'User-Agent': 'Mozilla/5.0'
}

response = requests.get(url, headers=headers)

# Save directly to file
with open("cat.png", "wb") as f:
    f.write(response.content)

# Open and display
image = Image.open("cat.png")
display(image)
```

```python
from gradio_client import Client
from IPython.display import display, Image as IPImage

# Using a dead simple SD 1.5 space
client = Client("anzorq/dreambooth-stable-diffusion")

# This space only takes ONE input: the prompt string.
result = client.predict(
    "A white siamese cat",
    api_name="/predict"
)

# The result here should be just a file path string
image_path = result if isinstance(result, str) else result[0]


print(f"Attempting to display: {image_path}")
display(IPImage(filename=image_path))
```
