```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
import getpass
```

```python
def set_if_undefined(var:str):
    if os.environ.get(var):
        return
    os.environ[var] = getpass.getpass(var)
set_if_undefined("GROQ_API_KEY")    
```

```python
def llm_model(prompt_txt, params = None): 
    """
    Create and invoke a Groq LLM with the given prompt and parameters.
    
    Args:
        prompt_txt (str): The prompt to send to the model
        params (dict): Optional parameters to customize model behavior
    
    Returns:
        str: The model's response
    """
    model_id = "llama-3.3-70b-versatile"
    default_parameters= {
        "temperature": 0.5, 
        "top_p" : 0.2, 
        "max_tokens": 256, 
     
    }
    if params:
        default_parameters.update(params)
    groq_model =  ChatGroq(
        model = model_id,
        temperature = default_parameters["temperature"], 
        top_p = default_parameters["top_p"], 
        
    )
    response = groq_model.invoke(prompt_txt)
    return response.content
    
```

```python
#Showing a demo of above function
prompt = "Tell me a joke about Seattle"
response = llm_model(prompt) 
print(f" The response is {response}")
```

```python
def joke_chain(prompt_txt, params = None): 
    """
    Create and invoke a Groq LLM with the given prompt and parameters and also creating a chaing using LCEL
    
    Args:
        prompt_txt (str): The prompt to send to the model
        params (dict): Optional parameters to customize model behavior
    
    Returns:
        str: The model's response
    """
    model_id = "llama-3.3-70b-versatile"
    default_parameters= {
        "temperature": 0.5, 
        "top_p" : 0.2, 
        "max_tokens": 256, 
     
    }
    if params:
        default_parameters.update(params)
    groq_model =  ChatGroq(
        model = model_id,
        temperature = default_parameters["temperature"], 
        top_p = default_parameters["top_p"], 
        
    )
    prompt = PromptTemplate.from_template(prompt_txt)
    parser = StrOutputParser() 
    chain = prompt| groq_model | StrOutputParser()
    return chain
```

```python
prompt1 = """Tell me a {adjective} joke about {content}"""
chain = joke_chain(prompt1)
response = chain.invoke({"adjective":"funny", "content":"Monroe"})
response
```

```python
# Text summarization
content = """
    The rapid advancement of technology in the 21st century has transformed various industries, including healthcare, education, and transportation. 
    Innovations such as artificial intelligence, machine learning, and the Internet of Things have revolutionized how we approach everyday tasks and complex problems. 
    For instance, AI-powered diagnostic tools are improving the accuracy and speed of medical diagnoses, while smart transportation systems are making cities more efficient and reducing traffic congestion. 
    Moreover, online learning platforms are making education more accessible to people around the world, breaking down geographical and financial barriers. 
    These technological developments are not only enhancing productivity but also contributing to a more interconnected and informed society.
"""
summary_template = """ Summarize the {content} in a sentence """ 
summary = PromptTemplate.from_template(summary_template)
summary_chain = summary | groq_model | StrOutputParser() 
response = summary_chain.invoke({"content":content}) 
print(response)
```

```python
# Question and Answer agent 
content = """
        The solar system consists of the Sun, eight planets, their moons, dwarf planets, and smaller objects like asteroids and comets. 
        The inner planets—Mercury, Venus, Earth, and Mars—are rocky and solid. 
        The outer planets—Jupiter, Saturn, Uranus, and Neptune—are much larger and gaseous.
"""

question = "Which planets in the solar system are rocky and solid?"
template = """ Answer the {question} based on {content}.
                Respond with "I don't know" if you are unsure.
                Answer : 

"""
prompt = PromptTemplate.from_template(template)
q_a_chain = prompt | groq_model | StrOutputParser() 
response = q_a_chain.invoke({"question":question, "content": content})
response
```

```python
#Text classification
text = """
        The concert last night was an exhilarating experience with outstanding performances by all artists.
"""

categories = "Entertainment, Food and Dining, Technology, Literature, Music."

template = """
        Classify the {text} into one of the {categories}.
        Reply with "Unable to decide" if you are unsure.
        Give the answer in this format 
        Category:

"""
prompt = PromptTemplate.from_template(template)
text_classification_chain = prompt | groq_model | StrOutputParser() 
response = text_classification_chain.invoke({"text":text, "categories": categories})
response
```

```python
#Role playing
role = """
        game master
"""

tone = "engaging and immersive"

template = """
            You are an expert {role}. I have this question {question}. I would like our conversation to be {tone}.
            
            Answer:
            
"""
prompt = PromptTemplate.from_template(template)
role_play_chain = prompt | groq_model | StrOutputParser()
```

```python
# SImulating a chatbot for above 
while True :
    query = input("Question: ")
    if query.lower() in ["quit", "exit", "bye"]:
        print("Answer: Goodbye")
        break
    response = role_play_chain.invoke({"role": role, "tone": tone , "question": query})
    print(f"Answer:{response}")
    
```
