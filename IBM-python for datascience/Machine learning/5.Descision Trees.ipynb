```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn import metrics
%matplotlib inline
```

```python
import requests
def download_file(url, file_name):
    response = requests.get(url)
    if response.status_code == 200: 
        with open(file_name, "wb") as f :
            f.write(response.content)
```

```python
path= 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%203/data/drug200.csv'
download_file(path, "drugs200.csv")
file_name = "drugs200.csv"
```

```python
my_data = pd.read_csv(file_name)
my_data.head()
```

```python
my_data.info()
```

```python
label_encoder = LabelEncoder()
my_data["Sex"] = label_encoder.fit_transform(my_data["Sex"])
my_data["BP"] = label_encoder.fit_transform(my_data["BP"])
my_data["Cholesterol"] = label_encoder.fit_transform(my_data["Cholesterol"])
my_data
```

```python
my_data.isnull().sum()
```

```python
custom_map = {'drugA':0,'drugB':1,'drugC':2,'drugX':3,'drugY':4}
my_data['Drug_num'] = my_data['Drug'].map(custom_map)
my_data
```

```python
my_data.drop('Drug',axis=1).corr()['Drug_num']
```

```python
category_counts = my_data['Drug'].value_counts()

# Plot the count plot
plt.bar(category_counts.index, category_counts.values, color='blue')
plt.xlabel('Drug')
plt.ylabel('Count')
plt.title('Category Distribution')
plt.xticks(rotation=45)  # Rotate labels for better readability if needed
plt.show()
```

```python
y = my_data["Drug"]
X = my_data.drop(["Drug", "Drug_num"], axis = 1)
```

```python
X_train, X_test , y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 42 )
```

```python
drugTree = DecisionTreeClassifier(criterion = "entropy" , max_depth = 3)
```

```python
drugTree.fit(X_train, y_train)
```

```python
drugTree.score(X_test, y_test)
```

```python
tree_predictions = drugTree.predict(X_test)
```

```python
print("Decision Trees's Accuracy: ", metrics.accuracy_score(y_test, tree_predictions))
```

```python
plot_tree(drugTree)
plt.show()
```
