```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import requests
def download_file(url,file_name):
    response = requests.get(url)
    if response.status_code == 200:
        with open(file_name, "wb") as f:
            f.write(response.content)
```

```python
file_path="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-SkillsNetwork/labs/Data%20files/auto.csv"
```

```python
download_file(file_path, "usedcars.csv")
file_name = "usedcars.csv"
```

```python
df = pd.read_csv(file_name)
df.head()
```

```python
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]
df = pd.read_csv(file_name, names = headers)
df.head()
```

```python
#convert ? to Nan
import numpy as np
df.replace("?",np.nan,inplace = True)
df.head()
```

```python
missing_data = df.isnull()
missing_data.head()
```

```python
#Count missing values in each column
for column in missing_data.columns.values.tolist():
    print(column)
    print(missing_data[column].value_counts())
    print("")
```

```python
#Calculate mean value of "normalized-losses column"
avg_norm_loss = df["normalized-losses"].astype("float").mean()
print("Average of normalized-losses:", avg_norm_loss)
```

```python
df["normalized-losses"].replace(np.nan, avg_norm_loss, inplace = True)
df.head()
```

```python
#calculate mean value for "bore column"
avg_bore = df["bore"].astype("float").mean()
df["bore"] = df["bore"].replace(np.nan, avg_bore)
df.head(25)
print("Average bore is :",avg_bore)
```

```python
#Calculating mean and replacing nan with it in "stroke" column
avg_stroke = df["stroke"].astype("float").mean()
df["stroke"] = df["stroke"].replace(np.nan, avg_stroke)
df.head()
print(avg_stroke)
```

```python
df[["bore", "stroke"]] = df[["bore", "stroke"]].astype("float")
df[["normalized-losses"]] = df[["normalized-losses"]].astype("int")
df[["price"]] = df[["price"]].astype("float")
df[["peak-rpm"]] = df[["peak-rpm"]].astype("float")
```

```python
df.dtypes
#plt.plot(df["peak-rpm"], df["price"])
plt.scatter(df["peak-rpm"], df["price"])
```

```python
avg_horsepower = df['horsepower'].astype('float').mean(axis=0)
print("Average horsepower:", avg_horsepower)
df['horsepower'].replace(np.nan, avg_horsepower, inplace=True)
```

```python
%matplotlib inline
import matplotlib as plt
from matplotlib import pyplot
df["horsepower"]=df["horsepower"].astype(int, copy=True)
plt.pyplot.hist(df["horsepower"])

# set x/y labels and plot title
plt.pyplot.xlabel("horsepower")
plt.pyplot.ylabel("count")
plt.pyplot.title("horsepower bins")
```

```python
bins = np.linspace(min(df["horsepower"]), max(df["horsepower"]),4)
bins
```

```python
group_names = ['Low', 'Medium', 'High']
```

```python
df["horsepower-binned"] = pd.cut(df["horsepower"], bins ,labels = group_names,include_lowest = True)
df[["horsepower", "horsepower-binned"]].head(20)
```

```python
df["horsepower-binned"].value_counts()
```

```python
%matplotlib inline
import matplotlib as plt
from matplotlib import pyplot
pyplot.bar(group_names,df["horsepower-binned"].value_counts())
plt.pyplot.xlabel("horsepower")
plt.pyplot.ylabel("count")
plt.pyplot.title("horsepower bins")
```
