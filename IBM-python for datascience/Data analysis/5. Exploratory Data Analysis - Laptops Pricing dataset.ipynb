```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
import requests
def download_file(url, file_name):
    response = requests.get(url)
    if response.status_code == 200:
        with open(file_name, "wb") as f:
            f.write(response.content)
```

```python
filepath="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/laptop_pricing_dataset_mod2.csv"
```

```python
download_file(filepath, "laptops.csv")
file_name = "laptops.csv"
```

```python
#Group the parameters "GPU", "CPU_core" and "Price" to make a pivot table and visualize this connection using the pcolor plot.
group1 = df[["GPU", "CPU_core", "Price"]]
group1_result = group1.groupby(["GPU", "CPU_core"], as_index = False).mean()
group1_result
```

```python
group1_result_pivot = group1_result.pivot(index = "GPU", columns = "CPU_core")
group1_result_pivot
```

```python
fig, ax = plt.subplots()
im = ax.pcolor(group1_result_pivot, cmap = "RdBu")
row_labels = group1_result_pivot.columns.levels[1]
col_labels = group1_result_pivot.index
#move ticks and labels to the center
ax.set_xticks(np.arange(group1_result_pivot.shape[1]) + 0.5, minor=False)
ax.set_yticks(np.arange(group1_result_pivot.shape[0]) + 0.5, minor=False)

#insert labels
ax.set_xticklabels(row_labels, minor=False)
ax.set_yticklabels(col_labels, minor=False)

fig.colorbar(im)
```

```python
#Pearson Correlation and p-values
from scipy import stats
for param in ['RAM_GB','CPU_frequency','Storage_GB_SSD','Screen_Size_inch','Weight_pounds','CPU_core','OS','GPU','Category']:
    pearson_coeff, p_value = stats.pearsonr(df[param],df["Price"])
    print(param)
    print("The Pearson Correlation Coefficient for ",param," is", pearson_coeff, " with a P-value of P =", p_value)
```
