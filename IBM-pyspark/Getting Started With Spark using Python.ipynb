```python
import findspark
findspark.init()
```

```python
from pyspark import SparkContext, SparkConf
from pyspark.sql import SparkSession
```

```python
# Creating sparkcontext
sc = SparkContext()
 #Creating a spark session
spark = SparkSession \
        .builder \
        .appName("Python spark Dataframes basic-example") \
        .config("spark.some.config.option", "somevalue") \
        .getOrCreate()
```

```python
if 'spark' in locals() and isinstance(spark , SparkSession):
    print("Spark session is active and ready to use")
else:
    print("Spark session isn't active.Please create a spark session.")
```

```python
data = range(1,30)
print(data[0])
len(data)
xrangeRDD = sc.parallelize(data,4)
xrangeRDD
```

```python
subRDD = xrangeRDD.map(lambda x: x-1)
filterRDD = subRDD.filter(lambda x: x<10)
```

```python
print(filterRDD.collect())
filterRDD.count()
```

```python
import time
test = sc.parallelize(range(1,5000),4)
test.cache()
t1 = time.time()
# first count will trigger evaluation of count *and* cache
count1 = test.count()
dt1 = time.time() - t1
print("dt1: ", dt1)


t2 = time.time()
# second count operates on cached data only
count2 = test.count()
dt2 = time.time() - t2
print("dt2: ", dt2)

test.count()
```
