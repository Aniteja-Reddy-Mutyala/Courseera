```python
import boto3
import json

# Use the ID from your active list
MODEL_ID_NOVA = "amazon.nova-lite-v1:0"

bedrock = boto3.client(service_name="bedrock-runtime", region_name="us-east-1")

# Nova expects the 'messages' format
payload = {
    "messages": [
        {
            "role": "user",
            "content": [{"text": "Explain quantum computing in simple terms?"}]
        }
    ],
    "inferenceConfig": {
        "max_new_tokens": 500,
        "temperature": 0.5,
        "topP": 0.9
    }
}

response = bedrock.invoke_model(
    modelId=MODEL_ID_NOVA,
    body=json.dumps(payload)
)

# Parse the response
response_body = json.loads(response['body'].read())
output_text = response_body['output']['message']['content'][0]['text']

print(output_text)
```

```python
# 1. Use the streaming method
response = bedrock.invoke_model_with_response_stream(
    modelId=MODEL_ID_NOVA,
    body=json.dumps(payload)
)

# 2. Iterate over the stream
for event in response.get("body"):
    chunk = json.loads(event.get("chunk").get("bytes"))
    
    # Nova's stream format: look for 'contentBlockDelta'
    if "contentBlockDelta" in chunk:
        text_delta = chunk["contentBlockDelta"]["delta"]["text"]
        print(text_delta, end="", flush=True)
```
